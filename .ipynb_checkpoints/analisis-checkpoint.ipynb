```python
import pandas as pd

# Si está en la misma carpeta que la notebook:
df = pd.read_csv('CKD_NHANES_2021_2023.csv')

df.head()
df
```

```python
df = df.drop_duplicates()
df
```

```python
# Conteo y porcentaje de valores faltantes por variable
nulos = pd.DataFrame({
    'nulos': df.isnull().sum(),
    'porcentaje': (df.isnull().mean() * 100).round(2),
}).sort_values(by='porcentaje', ascending=False)

print(nulos)
```

```python
print(df['insulin_use'].unique())
```

```python
# 1. Verificamos la distribución original (conteo incluyendo nulos)
print("Distribución original:")
print(df['insulin_use'].value_counts(dropna=False))
```

```python
# 2. Creamos una nueva columna limpia o transformamos la existente:
# En NHANES: 1 = 'Sí', 2 = 'No', NaN = No aplica (no tiene diabetes)
# Mapeamos:
#  1 -> 1 (Toma insulina)
#  2 -> 0 (Tiene diabetes, pero no toma insulina)
#  NaN -> 0 (No tiene diabetes, por ende no toma insulina)
# Opción A: Variable binaria (0 y 1)
df['takes_insulin'] = df['insulin_use'].map({1: 1, 2: 0}).fillna(0).astype(int)
print("\nDistribución luego de la limpieza:")
print(df['takes_insulin'].value_counts())
```

```python
# Variables declaradas típicas (ej: diabetes, smoking, etc.)
cols_cuestionario = [c for c in ['diabetes_diagnosed', 'ever_smoked', 'education_level', 'diabetes_pills'] if c in df.columns]

for col in cols_cuestionario:
    # 7 y 9 son valores no informativos en NHANES -> los pasamos a NaN
    df[col] = df[col].replace({7: np.nan, 9: np.nan})
```

```python
print("Distribución original:")
print(df['bp_diastolic'].value_counts(ascending=True))
```

```python
# Reemplazar valores no plausibles por NaN en lugar de borrar la fila entera
if 'bp_systolic' in df.columns and 'bp_diastolic' in df.columns:
    # Caso diastólica en 0 o fuera de rango
    df.loc[(df['bp_diastolic'] <= 30) | (df['bp_diastolic'] > 150), 'bp_diastolic'] = np.nan
    df.loc[(df['bp_systolic'] <= 60) | (df['bp_systolic'] > 280), 'bp_systolic'] = np.nan
```
